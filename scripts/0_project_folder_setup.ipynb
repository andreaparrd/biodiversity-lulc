# Step 1 — Set the Project Folder as the Working Directory

Setting the `biodiversity_lulc_project` as working directory.

Reminder: before running the next code block, **don't forget to change the runtime back to R**`

```r
# Setting the working directory
my_project_folder <- "biodiversity_lulc_project"

# Build the full path
project_path <- file.path("/content/drive/MyDrive", my_project_folder)
```

```r
# Create the project folder if it does not exist yet
dir.create(project_path, showWarnings = FALSE)
```

```r
# Set it as the working directory
setwd(project_path)

getwd()
```
